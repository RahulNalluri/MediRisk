```python
import pandas as pd
import numpy as np

# Load all datasets
df_diabetes = pd.read_csv('../data/diabetes_clean.csv')
df_diabetes2 = pd.read_csv('../data/diabetes_dataset_clean.csv')
df_heart = pd.read_csv('../data/heart_clean.csv')

print("Datasets loaded!")
print("Diabetes:", df_diabetes.shape)
print("Diabetes2:", df_diabetes2.shape)
print("Heart:", df_heart.shape)
```

```python
# Extract key statistics from diabetes dataset
diabetic = df_diabetes[df_diabetes['Outcome'] == 1]
healthy = df_diabetes[df_diabetes['Outcome'] == 0]

diabetes_text = f"""
DIABETES DATASET INSIGHTS — Pima Indians Diabetes Study
Total Patients: {len(df_diabetes)}
Diabetic Patients: {len(diabetic)} ({len(diabetic)/len(df_diabetes)*100:.1f}%)
Non-Diabetic Patients: {len(healthy)} ({len(healthy)/len(df_diabetes)*100:.1f}%)

GLUCOSE LEVELS:
Average glucose in diabetic patients: {diabetic['Glucose'].mean():.1f} mg/dL
Average glucose in healthy patients: {healthy['Glucose'].mean():.1f} mg/dL
Diabetic threshold: 126 mg/dL
Patients with glucose above 126: {(df_diabetes['Glucose'] > 126).sum()} patients

BMI ANALYSIS:
Average BMI in diabetic patients: {diabetic['BMI'].mean():.1f}
Average BMI in healthy patients: {healthy['BMI'].mean():.1f}
Obesity threshold BMI 30: {(df_diabetes['BMI'] > 30).sum()} patients are obese

AGE ANALYSIS:
Average age of diabetic patients: {diabetic['Age'].mean():.1f} years
Average age of healthy patients: {healthy['Age'].mean():.1f} years
Patients above 45 years: {(df_diabetes['Age'] > 45).sum()} patients

INSULIN LEVELS:
Average insulin in diabetic patients: {diabetic['Insulin'].mean():.1f} mu U/ml
Average insulin in healthy patients: {healthy['Insulin'].mean():.1f} mu U/ml

BLOOD PRESSURE:
Average blood pressure in diabetic patients: {diabetic['BloodPressure'].mean():.1f} mmHg
Average blood pressure in healthy patients: {healthy['BloodPressure'].mean():.1f} mmHg

KEY RISK FACTORS:
1. Glucose is the strongest predictor with 49% correlation to diabetes
2. BMI has 31% correlation to diabetes
3. Age has 24% correlation to diabetes
4. Pregnancies have 22% correlation to diabetes
5. Insulin has 20% correlation to diabetes

RISK ZONES:
High Risk: Glucose above 126 AND BMI above 30
Medium Risk: Glucose between 100-126 OR BMI between 25-30
Low Risk: Glucose below 100 AND BMI below 25
"""

# Save to file
with open('../backend/knowledge/diabetes.txt', 'w') as f:
    f.write(diabetes_text)

print("Diabetes knowledge base saved!")
print(diabetes_text)
```

```python
type_labels = {0: 'Prediabetic', 1: 'Type 1', 2: 'Type 2', 3: 'Type 3'}

diabetes2_text = f"""
DIABETES TYPES DATASET INSIGHTS — 70,000 Patient Study
Total Patients: {len(df_diabetes2)}

TYPE DISTRIBUTION:
{chr(10).join([f"{type_labels[i]}: {(df_diabetes2['Target']==i).sum()} patients ({(df_diabetes2['Target']==i).sum()/len(df_diabetes2)*100:.1f}%)" for i in range(4)])}

BLOOD GLUCOSE BY TYPE:
{chr(10).join([f"{type_labels[i]}: Average {df_diabetes2[df_diabetes2['Target']==i]['Blood Glucose Levels'].mean():.1f} mg/dL" for i in range(4)])}

AGE BY TYPE:
{chr(10).join([f"{type_labels[i]}: Average age {df_diabetes2[df_diabetes2['Target']==i]['Age'].mean():.1f} years" for i in range(4)])}

BMI BY TYPE:
{chr(10).join([f"{type_labels[i]}: Average BMI {df_diabetes2[df_diabetes2['Target']==i]['BMI'].mean():.1f}" for i in range(4)])}

INSULIN BY TYPE:
{chr(10).join([f"{type_labels[i]}: Average insulin {df_diabetes2[df_diabetes2['Target']==i]['Insulin Levels'].mean():.1f} mu U/ml" for i in range(4)])}

TYPE DESCRIPTIONS:
Prediabetic: Blood sugar higher than normal but not yet diabetic. Reversible with lifestyle changes.
Type 1: Immune system destroys insulin producing cells. Requires insulin injections for life.
Type 2: Body does not use insulin properly. Managed with diet exercise and medication.
Type 3: Diabetes caused by organ damage to pancreas or liver. Requires immediate medical attention.

MODEL PERFORMANCE:
XGBoost model accuracy on diabetes type classification: 95.07%
Prediabetic detection recall: 100%
Type 1 detection recall: 99%
Type 2 detection recall: 90%
Type 3 detection recall: 90%
"""

with open('../backend/knowledge/diabetes_types.txt', 'w') as f:
    f.write(diabetes2_text)

print("Diabetes types knowledge base saved!")
print(diabetes2_text)
```

```python
heart_disease = df_heart[df_heart['num'] == 1]
heart_healthy = df_heart[df_heart['num'] == 0]

heart_text = f"""
HEART DISEASE DATASET INSIGHTS — UCI Heart Disease Study
Total Patients: {len(df_heart)}
Heart Disease Patients: {len(heart_disease)} ({len(heart_disease)/len(df_heart)*100:.1f}%)
Healthy Patients: {len(heart_healthy)} ({len(heart_healthy)/len(df_heart)*100:.1f}%)

AGE ANALYSIS:
Average age of heart disease patients: {heart_disease['age'].mean():.1f} years
Average age of healthy patients: {heart_healthy['age'].mean():.1f} years
Patients above 55 years with heart disease: {((df_heart['age'] > 55) & (df_heart['num'] == 1)).sum()} patients

GENDER ANALYSIS:
Male patients with heart disease: {((df_heart['sex'] == 1) & (df_heart['num'] == 1)).sum()} patients
Female patients with heart disease: {((df_heart['sex'] == 0) & (df_heart['num'] == 1)).sum()} patients

CHOLESTEROL ANALYSIS:
Average cholesterol in heart disease patients: {heart_disease['chol'].mean():.1f} mg/dL
Average cholesterol in healthy patients: {heart_healthy['chol'].mean():.1f} mg/dL
High cholesterol above 240: {(df_heart['chol'] > 240).sum()} patients

BLOOD PRESSURE ANALYSIS:
Average blood pressure in heart disease patients: {heart_disease['trestbps'].mean():.1f} mmHg
Average blood pressure in healthy patients: {heart_healthy['trestbps'].mean():.1f} mmHg
High blood pressure above 140: {(df_heart['trestbps'] > 140).sum()} patients

BLOCKED VESSELS ANALYSIS:
Patients with 0 blocked vessels: {(df_heart['ca'] == 0).sum()} patients
Patients with 1 blocked vessel: {(df_heart['ca'] == 1).sum()} patients
Patients with 2 blocked vessels: {(df_heart['ca'] == 2).sum()} patients
Patients with 3 blocked vessels: {(df_heart['ca'] == 3).sum()} patients
Heart disease rate with 3 blocked vessels: {(df_heart[df_heart['ca']==3]['num'].mean()*100):.1f}%

KEY RISK FACTORS:
1. Blocked vessels strongest predictor with 45.6% correlation
2. ST depression has 38.6% correlation to heart disease
3. Age has 28.3% correlation to heart disease
4. Blood pressure has 10.6% correlation
5. Cholesterol has 23.1% correlation

EXERCISE PAIN ANALYSIS:
Patients with exercise pain and heart disease: {((df_heart['exang'] == 1) & (df_heart['num'] == 1)).sum()} patients
Heart disease rate with exercise pain: {(df_heart[df_heart['exang']==1]['num'].mean()*100):.1f}%
Heart disease rate without exercise pain: {(df_heart[df_heart['exang']==0]['num'].mean()*100):.1f}%

MODEL PERFORMANCE:
XGBoost model accuracy on heart disease detection: 92.93%
Heart disease detection precision: 98%
Heart disease detection recall: 90%
Healthy patient detection recall: 97%
"""

with open('../backend/knowledge/heart_disease.txt', 'w') as f:
    f.write(heart_text)

print("Heart disease knowledge base saved!")
print(heart_text)
```
